ALGORITHM:1- Linear Simple regression

```python
import numpy as np
import random
import matplotlib.pyplot as plt
import seaborn as sns

#Step-1.To create tabular form dataset containing random input datas.
np.random.seed(30)
age=list()
networth=list()
for i in range(100):
  age.append(np.random.randint(20,65))
for j in age:
  networth.append(6.25 * j + np.random.normal(scale=40))
print(age)
print(networth)
A=np.array(age)
Nw=np.array(networth)
print(A)
print(Nw)
ages=A.reshape(len(age),1)
networths=Nw.reshape(len(networth),1)
print(ages,networths)

#Step-2.To split or divide whole input dataset into two parts which will be Training and Testing Datasets.
#train_test_split is a function in Sklearn model selection for splitting data arrays into two subsets: for training data and for testing data. With this
#function, you don't need to divide the dataset manually.And it returns the dataset divided in four parts which need to get stored in four variables.
from sklearn.model_selection import train_test_split
ages_train, ages_test, networths_train, networths_test = train_test_split(ages,networths)

#Step-3.TRAINING SESSION--
#A)Giving and Training the model with training dataset.
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(ages_train,networths_train)
#B)Now we visualize how training dataset train our model by plotting scatter plot with analyser line.
sns.regplot(x=ages_train,y=networths_train,scatter=True,color='b',marker='*')
plt.xlabel('Ages Train')
plt.ylabel('NetWorths Train')
plt.title("Regression Plot")
plt.figure(figsize=(12,10))
plt.show()

#Step-4.TESTING SESSION--
#A)Testing the model with testing dataset of only independent variable(i.e.ages_test).
print("Slope:",reg.coef_)
print("Intercept",reg.intercept_)
networths_pred=reg.predict(ages_test)
print(networths_pred)
#B)Now we visualize how predictor line is formed when only x axis testing data is given i.e how it predicts y-axis data.
plt.plot(ages_test,networths_pred)
plt.xlabel('Ages Test')
plt.ylabel('NetWorths Pred')
plt.title("Regression Testing Plot")
plt.figure(figsize=(12,10))
plt.show()

#Step-5.EFFICIENCY calculated by using r-squared score method.
#It internally calculates y_pred again on the basis of both ages_train and ages_test and gives the efficiency on comparing y_pred of
#both input datas with networths_train and networths_test
print("Testig data",reg.score(ages_test, networths_test))
print("Training data",reg.score(ages_train, networths_train))

#Step-6.Now we visualize the predictor line on both Training dataset and Testing dataset by plotiing graph including whole input dataset.
#A) Scatter plots are on the basis of already defined Training and Testing Datasets
plt.scatter(ages_train,networths_train,color='g',label='Training Dataset')
plt.scatter(ages_test,networths_test,color='r',label='Testing Dataset')
#This funct is used to tell that where we want to show these labels associated/represented by colours in our graph i.e.define their location in graph
plt.legend(loc=2)
#B) Predictor line drawn on The basis of networth predicted data on given ages testing data.
plt.plot(ages_test,networths_pred)

plt.xlabel('Ages')
plt.ylabel('NetWorths')
plt.figure(figsize=(12,10))
plt.show()
```

ALGORITHM:2- K-Nearest Neighbours

```python
#Step-1.To define input values so to get tabular dataset of 2 columns containing x which is other var & y which is categorial var.
#X contains 0,1,2,3 named mails which belongs to inbox,inbox,spam,spam respective classes stored in var Y.
X = [[0],[1],[2],[3]]
Y = ['inbox','inbox','spam','spam']

#Step-2.TRAINING SESSION--
#Giving and Training the model with training dataset.First import Model from sklearn,then create object of model class while calling the parametrized const
#by passing argument of keyword which give the value of K i.e.how many no of nearest neighbors we want to consider for unknown pt.And finally Training the
#model with training data stored in X & Y var on basis of which it form scatterplot graph containing plots of y value on given x var values internally.
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier(n_neighbors=3)
knn.fit(X,Y)

#Step-3.TESTING SESSION--
#For a classification model, the predicted class for each sample in X is returned.
#Testing the model with testing dataset.First define new datatset containing different new mails(i.e different unknown class plots) which will be testing
#dataset for model,then let model predict the class or category to which each mail belongs to (i.e.class to which each given new plot belongs to)
Z = [[0.9],[1.1],[1.5],[3.3]]
print(knn.predict(Z))
#So classes are predicted respectively for different unknown points.Seeing our training data,for 0pt class defined is'inbox',for 1pt class defined is again
#'inbox'and so on.Now on seeing testing data,the first and second new pts i.e.0.9, 1.1 & 1.5 are close to 0,1,2 pts as it considers 3 nearest neighbor plots so
#the most frequently repeating class of nearest plot around these new pts are 'inbox'class, thus new pts also get'inbox'class and so on.

#Step-4.EFFICIENCY calculated by using Accuracy Classification Score method.
#This imported model is in form of function which is only use to calculate efficiency for Classification technique.So we can directly pass the actual test and
#predicted values in funct.
from sklearn.metrics import accuracy_score
#Example.1
giv=['spam','spam']
pre=knn.predict([[2],[3]])
print(pre)
print(accuracy_score(giv,pre))
#Example.2
y_pred=[0,2,1,3]
y_actual=[0,1,2,3]
print(accuracy_score(y_actual,y_pred))
#So it compares each value of actual testing data to its corresponding value of predicted data(like:0 & 0,2 & 1,...) and give efficiency on the basis of the
#values which are matching with each other(like here:first and last valuesi.e 0&0,3&3 are matching with each other on comparing so it gives 50% efficiency)
```

ALGORITHM:3- Decision Tree

```python
#Step-1.To define input values so to get tabular dataset of 2 columns containing x which is other var & y which is categorial var.
#X contains [0,0],[1,1] datas which belongs to Class 0 and Class 1 respectively stored in Y var.
X = [[0,0],[1,1]]
Y = [0,1]

#Step-2.TRAINING SESSION--
#Giving and Training the model with training dataset.First import Model from sklearn,then create object of model class. And finally Training the model with
#training data stored in X & Y var on basis of which it arranges data in form of tree internally which gives all possible paths which ultimately leading to
#some class.
from sklearn.tree import DecisionTreeClassifier
dtre=DecisionTreeClassifier()
dtre.fit(X,Y)

#Step-3.TESTING SESSION--
#Testing the model with testing dataset.First define new datatset containing new value only for other var which will be testing dataset for model,then let
#model predict the class or category to which that testing data belongs to i.e.predict the value of categorial variable.
dtre.predict([[2,2]])
#For a classification model, the predicted class for each sample in X is returned.
```

ALGORITHM:5- Random Decision Forest

PART I:*Random Forest Regressor*

```python
#Step-1.To import tabular form of dataset already contains random data.
#make_regression is a function imported from Sklearn datasets which gives random regression dataset with 100 no of rows and 4 no of columns among which 2 columns
#are informative,so the no of informative features means no of independent variables which used to build(train)the linear model and on the basis of this only the
#output is generated in dependent var with shuffle as False becaz bydefault everytime it execute it shuffle the samples and the features. And also random_state
#which determines random number generation for dataset creation,so pass an int value 0 so that there will be no reproducible output(i.e.same random data are
#generated everytime) across multiple function calls..And it returns the input dataset of given rows and col which we store/assign as independent vars in X &
#output values equal to no of row stored as dependent var values based on each row of independent var in X.
from sklearn.datasets import make_regression
X,y=make_regression(n_samples=100,n_features=4,n_informative=2,random_state=0,shuffle=False)
print(X)
print(y)
print(X.shape)
print(y.shape)

#Step-2.To split or divide whole input dataset into two parts which will be Training and Testing Datasets.
#train_test_split is a function in Sklearn model selection for splitting data arrays into two subsets: for training data and for testing data. With this
#function, you don't need to divide the dataset manually.And it returns the dataset divided randomly in four parts which need to get stored in four variables
#and generate same random subsets each time when funct runs and splitting dataset into 85% of training data and 15% of testing data as test_size means testing
#data size.
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=0,test_size=0.15)

#Step-3.TRAINING SESSION--
#Giving and Training the model with training dataset.First import Model from sklearn,then create object of model class with passing arguments as no of levels
#in tree formed with same random values generated. And finally Training the model with training data stored in both training var on basis of which it arranges
#data in form of different trees for each different models internally i.e.training multiple same models ata time with different training subsets becaz of which
#it forms different decision trees for each model.
from sklearn.ensemble import RandomForestRegressor
regr = RandomForestRegressor(max_depth=2,random_state=0)
regr.fit(X_train,y_train)

#Step-4.TESTING SESSION--
#Testing the model with testing dataset.For a regression model,the predicted value based on X i.e.test data of independent var(X_test) is returned and thus it
#predict the value of dependent variable which we already have in y_test.
y_pred=regr.predict(X_test)
print(y_pred)

#Step-5.EFFICIENCY calculated by using r-squared score method.
#This imported model is in form of function which is only use to calculate efficiency for Regression technique.So we can directly pass the actual test and
#predicted values in funct.
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```

PART II:*Random Forest Classifier*

```python
#Step-1.To import tabular form of dataset already contains random data.
#make_regression is a function imported from Sklearn datasets which gives random classification dataset with 100 no of rows and 4 no of columns among which 2 columns
#are informative,so the no of informative features means no of other variables(consist all vars except categorial var)which used to build(train)this model and on
#the basis of this only the output is generated in categorial var with shuffle as False becaz bydefault everytime it execute it shuffle the samples & the features.
#And also random_state which determines random number generation for dataset creation,so pass an int value 0 so that there will be no reproducible output(i.e.same
#random data are generated everytime) across multiple function calls.And it returns the input dataset of given rows and col which we store/assign as the all other
#vars in X & output values equal to no of row stored as categorial var classes based on each row of other vars in X1.
from sklearn.datasets import make_classification
X1,y1=make_classification(n_samples=100,n_features=4,n_informative=2,random_state=0,shuffle=False)
print(X1)
print(y1)
print(X1.shape)
print(y1.shape)

#Step-2.To split or divide whole input dataset into two parts which will be Training and Testing Datasets.
#train_test_split is a function in Sklearn model selection for splitting data arrays into two subsets: for training data and for testing data. With this
#function, you don't need to divide the dataset manually.And it returns the dataset divided randomly in four parts which need to get stored in four variables
#and generate same random subsets each time when funct runs.
from sklearn.model_selection import train_test_split
X1_train, X1_test, y1_train, y1_test = train_test_split(X1,y1,random_state=0,test_size=0.15)

#Step-3.TRAINING SESSION--
#Giving and Training the model with training dataset.First import Model from sklearn,then create object of model class with passing arguments as no of levels
#in tree formed with same random values generated. And finally Training the model with training data stored in both training var on basis of which it arranges
#data in form of different trees for each different models internally i.e.training multiple same models ata time with different training subsets becaz of which
#it forms different decision trees for each model.
from sklearn.ensemble import RandomForestClassifier
clfr = RandomForestClassifier(max_depth=2,random_state=0,min_samples_split=3)
clfr.fit(X1_train,y1_train)

#Step-4.TESTING SESSION--
#Testing the model with testing dataset.For a classification model, the predicted class for each sample in X i.e.test data of other vars(X1_test)is returned and
#thus it predict the classes in categorial variable which we already have in y1_test.
y1_pred=clfr.predict(X1_test)
print(y1_pred)

#Step-5.EFFICIENCY calculated by using Accuracy Classification Score method.
#This imported model is in form of function which is only use to calculate efficiency for Classification technique.So we can directly pass the actual test and
#predicted values in funct.
from sklearn.metrics import accuracy_score
accuracy_score(y1_test,y1_pred)
```

ALGORITHM:6- K-Means Algorithm

```python
import numpy as np
#Step-1.To define input values so to get tabular dataset of 1 column.
#X contains [1,2],[1,4],[1,0],[10,2],[10,4],[10,0] pts which can be grouped in particular clusters,i.e.since clustering is unsupervised type of ML so we don't
#define clusters to which other columns data belong to as training data rather we only give all column values which are just input columns as training data and
#it will itself try to make clusters of that training data.Thus it does not need output column but only input columns as training dataset.
X = np.array([[1,2],[1,4],[1,0],[10,2],[10,4],[10,0]])
print(X)

#Step-2.TRAINING SESSION--
#Giving and Training the model with training dataset.First import Model from sklearn,then create object of model class while calling the parametrized const
#by passing argument of keyword which give the value of K i.e.how many no of random centre pts we want to consider which is equal to no of clusters formed.
#And finally Training the model with training data stored in X var on basis of which it form scatterplot graph containing plots of X values representing the
#positions internally.And thus run all steps of algorithm and group all the pts in k no of clusters having fixed centre pts..Random_state 0 is given so each
#time on execution it does not change the name of cluster for given pts i.e.name of part cluster to which par pts belongs to remain same.
#Thus it itself makes or defines clusters of training dataset.
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=2,random_state=0)
kmeans.fit(X)

#Step-3.TESTING SESSION--
#For a clustering model, the predicted cluster or group for each sample in X is returned.
#Testing the model with testing dataset.First define new datatset containing new different plots which will be testing dataset for model, then let model
#predict the cluster or group to which each given new plot belongs to acc to its position. We can also see clusters assigned/given to the training dataset
#pts(here two clusters i.e.cluster named 0 and 1 is formed)and the final centre pts i.e.centroid pos of each cluster(here two centre pts is defined which is
#fixed centre for two clusters) which is stored in member variables of model class.
#Thus according to clusters defined by it during Training it will decide cluster of testing dataset.
print('clusters assigned to each data pt',kmeans.labels_)
print("final centre pt of each cluster formed",kmeans.cluster_centers_)
print(kmeans.predict([[0,0],[12,3]]))
#So, it predict the closest cluster according to position of each new value.
```

ALGORITHM:7- Naive Bayes Algorithm

Bayes Formula:  
P(A/B) = P(B/A) * P(A)P(B)

where,

         P(A)= Probability of event A
         P(B)= Probability of event B
         P(A/B)= Probability of event A given event B has already occured
         P(B/A)= Probability of event B given event A has already occured

Naive Bayes Classifier is divided into three forms:
* Multinomial
* Gaussian
* Bernoulli's

ALGORITHM:8- AutoML using H2O-ML Framework

It is open-source ML Framework which provide various tested Statistical and ML Algorithms which can directly apply to Datasets.The H2O also provides web-based tool to test the different algorithms on your dataset,this is called Flow.

It also supports AutoML functionality which provides ranking based on the performance of several algorithms on dataset i.e.It give the highest performing ML Model on dataset.So,AutoML will identify the best performing or most efficient algorithm on your dataset.
Here we will see how AutoML is applied on dataset.


*Import Library*

```python
import h2o
from h2o.automl import H2OAutoML
```

*Initialize H2O*

```python
h2o.init()
```

*Steps to apply AutoML by using H2O Framework*

```python
#Step:1-
#LOADING DATASET
#A)First importing dataset family containing various datasets here in collaboratory from sklearn library
from sklearn import datasets
#B)Then get(to access)any particular datafile as dataset from imported datasets by using H2O library special funct.
data=h2o.import_file('https://gist.githubusercontent.com/btkhimsar/ed560337d8b944832d1c1f55fac093fc/raw/6f9306ad21398ea43cba4f7d537619d0e07d5ae3/iris.csv')

#PREPEARING DATASET
#We need to define input columns(i.e.var values on basis of which it predicts output) and output column(i.e.var whose value to be predicted).
print(data.columns)
inputft=['sepal.length','sepal.width','petal.length','petal.width']
outputft='variety'
```

```python
#Step:2-Data Preprocessing
#To split or divide whole input dataset into two parts which will be Training and Testing Datasets.Here no need to import some model to split data but we only
#call the funct by var containing dataset which is to be splitted and returns two subset where first is bydefault Training dataset. It split the data as given
#in 80:20 ratio for training and testing.
train,test = data.split_frame(ratios=[0.8])
```

```python
#Step:3-TRAINING SESSION
#Apply H2OAutoML model on our dataset and it give us the optimized model for given dataset.
#The first parameter specifies the number of models that we want to evaluate(finding efficiency)and compare on dataset.
#The second parameter specifies the time for which the algorithm runs, i.e.it will run for a fixed amount of time(in sec) set by us on dataset.
#The third parameter is to define that each time when autoML runs it will use that same type of models.
automl = H2OAutoML(max_models=30, max_runtime_secs=120, seed=1)

#Now the train member method is called by the object of H2OAutoML class so to train our H2OAutoML on basis of training data.
#Set it to the name of the column whose values we are trying to predict in our data.
automl.train(x=inputft, y=outputft, training_frame=train)
```

```python
#When AutoML processing completes,it creates a leaderboard ranking all the 30 algorithms that it has evaluated.To see the first 15 records of the leaderboard.
#The ranking details is stored in member var'leaderboard'of H2OAutoML class,so we can directly access it by using object of that class.
lb=automl.leaderboard
lb.head(15)
```

```python
#On accessing the var leaderboard it will print bydefault only first 10 rows.so,to see the ranks of all the tested algorithms(here 30) i.e.to print the whole
#leaderboard as it is(printing all 30 rows of ranking table).
lb.head(rows = lb.nrows)
```

```python
#step:4-TESTING SESSION
#Now, you have the models ranked on basis of training data,so then we can see the performance of the top-rated model on test data. Thus now we know which model
#is giving highest efficiency.Here automl will automatically apply that highest efficient model to testing dataset.
automl.predict(test)
#It print predicted result
```
